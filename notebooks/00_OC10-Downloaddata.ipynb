```python
from ftplib import FTP
import os
import re
import requests
```

```python
dataroot="data"
basepath="/pub/ITSG/GRACE/"
gfopath='ITSG-Grace_operational/monthly/monthly_n96/'
grcpath='ITSG-Grace2018/monthly/monthly_n96/'
```

```python
def get_grace(gpath):

    ftproot="ftp.tugraz.at"
    ftp=FTP(ftproot)
    ftp.login()



    ftp.cwd(os.path.join(basepath,gpath))

    datadir=os.path.join(dataroot,gpath)
    if not os.path.exists(datadir):
        os.makedirs(datadir,exist_ok=True)
                     
                     
    for f,_ in ftp.mlsd():
        fout=os.path.join(datadir,f)
        if not re.search('^ITSG.*gfc',f):
            continue
        if os.path.exists(fout):
            print(f"Skipping {f}")
        else:
            print(f"downloading {f}")
            with open(fout,'wb') as fid:
                ftp.retrbinary(f'RETR {f}' , fid.write)

```

```python
get_grace(grcpath)
```

```python
get_grace(gfopath)
```

```python
# also retrieve a static gravity field to subtract
icgemurl="https://icgem.gfz.de/getmodel/gfc/32ec2884630a02670476f752d2a2bf1c395d8c8d6d768090ed95b4f04b0d5863/GOCO06s.gfc"
icgemf=os.path.join(dataroot,os.path.basename(icgemurl))


if not os.path.exists(icgemf):
    req=requests.get(icgemurl)

    with open(icgemf,'wb') as fid:
        fid.write(req.content)




```
